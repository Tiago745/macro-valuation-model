# Importando valores da ETF

```python
import pandas as pd
import yfinance as yf

ticker_symbol = "EWZ"

# Baixar dados mensais diretamente
data = yf.download(
    tickers=ticker_symbol,
    start="2015-01-01",
    end="2025-01-01",
    interval="1mo",
    auto_adjust=False  # garante que Adj Close vai existir
)

# Se "Adj Close" existir, ok. Se não, cria copiando "Close".
if "Adj Close" not in data.columns:
    data["Adj Close"] = data["Close"]


# Resetar índice para transformar Date em coluna
data = data.reset_index()


# Salvar CSV
data.to_csv(f"ETF/{ticker_symbol}_1mo_data.csv")

print(f"CSV salvo como {ticker_symbol}_1mo_data.csv")
print(data.tail())
```
